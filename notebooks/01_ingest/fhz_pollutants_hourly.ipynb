# Merging

```python
import re

def normalize(col):
    # lowercase, strip, collapse all whitespace/punctuation differences
    col = col.strip().lower()
    col = re.sub(r'[\s\._]+', '', col)   # remove spaces, dots, underscores entirely
    return col

CANONICAL_MAP = {
    'starttime': 'start_time',
    'endtime': 'end_time',
    'pm101h': 'pm10_1h',
    'pm10h': 'pm10_1h',
    'pm10': 'pm10_1h',
    'pm251h': 'pm25_1h',
    'pm25': 'pm25_1h',
    'so21h': 'so2_1h',
    'so2': 'so2_1h',
    'no21h': 'no2_1h',
    'no2': 'no2_1h',
    'o31h': 'o3_1h',
    'o3': 'o3_1h',
    'co1h': 'co_1h',
    'co': 'co_1h',
}

def standardize_columns(df):
    new_cols = {}
    for col in df.columns:
        norm = normalize(col)
        new_cols[col] = CANONICAL_MAP.get(norm, norm)  # fall back to normalized name if unknown
    df = df.rename(columns=new_cols)

    if "start_time" in df.columns:
        df = df.drop(columns=["start_time"])

    return df
```
